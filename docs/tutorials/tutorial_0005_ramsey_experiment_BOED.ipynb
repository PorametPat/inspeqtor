## Define control

Let's over engineer our control sequence in this example. Since, we use a simulator, we can "cheat" by initialize a quantum state into $|+\rangle$ and measurement Pauli-X directly. However, let us not cheat and do it like a giga chad 💪. We first define the atomic control, a controllable delayed $\pi / 2$  rotation around X-axis at the start and end of the sequence.

```python
from dataclasses import dataclass


@dataclass
class DelayedX(sq.control.BaseControl):
    max_idle_time: float
    qubit_drive_strength: float
    dt: float
    max_amp: float = 0.25
    theta: float = jnp.pi / 2
    offset: float = 0.0

    def __post_init__(self):
        # This is the correction factor that will cancel the factor in the front of hamiltonian
        self.correction = 2 * jnp.pi * self.qubit_drive_strength * self.dt

        # The standard derivation of Gaussian pulse is keep fixed for the given max_amp
        self.sigma = jnp.sqrt(2 * jnp.pi) / (self.max_amp * self.correction)

    def get_bounds(
        self,
    ) -> tuple[sq.control.ParametersDictType, sq.control.ParametersDictType]:
        lower: sq.control.ParametersDictType = {
            "t": 0.0,
        }
        upper: sq.control.ParametersDictType = {
            "t": self.max_idle_time,
        }

        return lower, upper

    def get_envelope(self, params: sq.control.ParametersDictType):
        # The area of Gaussian to be rotate to,
        area = (
            self.theta / self.correction
        )  # NOTE: Choice of area is arbitrary e.g. pi pulse
        center_position = params["t"] + self.offset

        return sq.control.library.gaussian_envelope(
            amp=area, center=center_position, sigma=self.sigma
        )


total_dt = 1000
dt = 0.5

qubit_info = sq.data.library.get_mock_qubit_information()

ctrl_seq = sq.control.ControlSequence(
    controls={
        "start": DelayedX(
            max_idle_time=total_dt,
            qubit_drive_strength=qubit_info.drive_strength,
            dt=dt,
            max_amp=1.0,
            offset=50.0,
        ),
        "end": DelayedX(
            max_idle_time=total_dt,
            qubit_drive_strength=qubit_info.drive_strength,
            dt=dt,
            max_amp=1.0,
            offset=-50.0,
        ),
    },
    total_dt=total_dt,
)
ravel_fn, unravel_fn = sq.control.ravel_unravel_fn(ctrl_seq.get_structure())
```

We can visualize our lovely pulse sequence as usual.

## Prepare the simulator

In this case, we would like to have a fine-grained control over the simulator, so we are going to define a custom transmon Hamiltonain in rotating frame. The part that needed beyond the predefined version is the "detune" free evolution of the qubit. In this case, we turn on the detune at $100 < t < T_{\rm end} - 100$.

```python
import typing


def rotating_transmon_hamiltonian(
    params,
    t: jnp.ndarray,
    qubit_info: sq.data.QubitInformation,
    signal: typing.Callable[..., jnp.ndarray],
) -> jnp.ndarray:
    a0 = 2 * jnp.pi * qubit_info.frequency
    a1 = 2 * jnp.pi * qubit_info.drive_strength

    def f_sigma_x(params, t):
        return a1 * signal(params, t) * jnp.cos(a0 * t)

    def f_sigma_y(params, t):
        return a1 * signal(params, t) * jnp.sin(a0 * t)

    def piecewise(params, t):
        detune_freq = 0.005

        return jnp.piecewise(
            t, [t < 100, t > params["end"]["t"] - 100], [0.0, 0.0, detune_freq]
        )

    return (
        f_sigma_x(params, t) * sq.utils.X
        - f_sigma_y(params, t) * sq.utils.Y
        + a0 * 0.5 * sq.utils.Z * piecewise(params, t)
    )


def shot_quantum_device(
    key: jnp.ndarray,
    param,
    initial_state: jnp.ndarray,
    observable: jnp.ndarray,
    shots: int,
    hamiltonian,
    lindblad_ops,
):
    # Solver is a function that return the final density matrix in this case
    # calculate the final density matrix
    t_eval = jnp.array([param["start"]["t"], param["end"]["t"]])
    dm = sq.physics.lindblad_solver(
        args=param,
        t_eval=t_eval,
        hamiltonian=hamiltonian,
        lindblad_ops=lindblad_ops,
        rho0=initial_state,
        t0=param["start"]["t"],  # type: ignore
        t1=param["end"]["t"],  # type: ignore
        max_steps=int(2**17),
    )[-1]

    return sq.utils.calculate_shots_expectation_value(
        key, dm, unitary=jnp.eye(2), operator=observable, shots=shots
    )
```

## Simulate the experiment.

Since, we perform an experiment that did not required sampling of the control sequence, we simply construct the control parameter for each sequence directly. Typically, for Ramsey experiment, we refer the decoherence time of the phase as $T_2^*$, but to keep it brief, we keep the variable name as `T2`. 

```python
key = jax.random.key(0)
initial_state = sq.data.get_initial_state("0", dm=True)

sample_size = 201
shots = 1000

t_end = jnp.linspace(300, total_dt, sample_size).reshape(-1, 1)
t_start = jnp.zeros_like(t_end).reshape(-1, 1)

t_evals = jnp.hstack([t_start, t_end])
# Indeed, we can vmap the flat parameters to the structured one.
t_evals = jax.vmap(unravel_fn)(t_evals)

# Define the device from functions above.
T2 = 500  # This is the value that we want to infer.
gamma_phi = 1 / T2  # Decay rate
L_dephase = jnp.sqrt(gamma_phi) * sq.utils.Z

drive_freq = qubit_info.frequency * 1.00
signal_fn = sq.physics.make_signal_fn(
    ctrl_seq.get_envelope,
    drive_frequency=drive_freq,
    dt=dt,
)

hamiltonian = partial(
    rotating_transmon_hamiltonian,
    qubit_info=qubit_info,
    signal=signal_fn,
)

device = partial(
    shot_quantum_device,
    initial_state=initial_state,
    observable=sq.utils.Z,
    shots=shots,
    hamiltonian=hamiltonian,
    lindblad_ops=[L_dephase],
)

# And we can vmap over the structured parameters too!
finite_shot_expvals = jax.vmap(device, in_axes=(None, 0))(key, t_evals)
```

```python
def recursive_vmap(
    fun,
    in_axes: tuple[int, ...] = (0,),
    out_axes: int | tuple[int, ...] = 0,
    base_ndims: int | tuple[int, ...] = 1,
):
    """
    Recursively applies jax.vmap, inferring the depth based on the
    number of dimensions of the arguments at call time.

    Args:
        fun: The base function to be mapped.
        in_axes: Specifies which axes to map over.
        out_axes: Specifies where the mapped axis should appear in the output.
        base_ndims: The number of core dimensions `fun` expects for its inputs.
                    Can be an int (applied to the first leaf) or a tuple of ints matching `args`.
    """

    def wrapper(*args, **kwargs):
        depth = None

        if isinstance(base_ndims, (tuple, list)):
            depths = []
            for arg, b_ndim in zip(args, base_ndims):
                if b_ndim is not None and hasattr(arg, "ndim"):
                    depths.append(arg.ndim - b_ndim)
            if depths:
                depth = depths[0]
                if not all(d == depth for d in depths):
                    raise ValueError(f"Inferred depths are inconsistent: {depths}")
        else:
            leaves = jax.tree_util.tree_leaves((args, kwargs))
            if leaves:
                depth = leaves[0].ndim - base_ndims

        if depth is None:
            return fun(*args, **kwargs)

        if depth < 0:
            raise ValueError(f"Calculated depth {depth} is less than 0.")

        mapped_fun = fun
        for _ in range(depth):
            mapped_fun = jax.vmap(mapped_fun, in_axes=in_axes, out_axes=out_axes)

        return mapped_fun(*args, **kwargs)

    return wrapper
```

```python
@jax.jit
def device_with_T2(T2, param):
    param = unravel_fn(param)
    gamma_phi = 1 / T2
    L_dephase = jnp.sqrt(gamma_phi) * sq.utils.Z

    # Solver is a function that return the final density matrix in this case
    # calculate the final density matrix
    t_eval = jnp.array([param["start"]["t"], param["end"]["t"]])
    dm = sq.physics.lindblad_solver(
        args=param,
        t_eval=t_eval,
        hamiltonian=hamiltonian,
        lindblad_ops=[L_dephase],
        rho0=initial_state,
        t0=param["start"]["t"],
        t1=param["end"]["t"],
        max_steps=int(2**17),
    )[-1]

    return jnp.real(jnp.trace(dm @ sq.utils.Z))
```

```python
import numpyro
import numpyro.distributions as dist

def probabilistic_model(t_evals):
    # The plate_stack handles BOTH the batched candidate designs and the particle dimension from BOED.
    with numpyro.plate_stack("plate", sizes=t_evals.shape[:-1]):
        # T2 gets shape (..., num_designs), treating each design evaluation as an independent experiment
        T2 = numpyro.sample("T2", dist.Normal(loc=500.0, scale=10.0))

        # We can map over the batch dimensions. Since T2 has the same batch shape as t_evals[:-1],
        # we can just vmap dynamically according to the batch dimensions we have.
        # But `recursive_vmap` with `base_ndims=(0, 1)` uses the batch rank perfectly!
        finite_shot_expvals = recursive_vmap(
            device_with_T2, in_axes=(0, 0), base_ndims=(0, 1)
        )(T2, t_evals)

        prob_minus = sq.utils.expectation_value_to_prob_minus(finite_shot_expvals)

        numpyro.sample("shots", dist.BernoulliProbs(probs=prob_minus))


t_end = jnp.linspace(300, total_dt, sample_size).reshape(-1, 1)
t_start = jnp.zeros_like(t_end).reshape(-1, 1)
t_evals = jnp.hstack([t_start, t_end])

trace = sq.models.probabilistic.get_trace(probabilistic_model)(t_evals)
trace
```

```python
def fast_probabilistic_model(t_evals):
    with numpyro.plate_stack("plate", sizes=t_evals.shape[:-1]):
        T2 = numpyro.sample("T2", dist.Normal(loc=500.0, scale=10.0))
        
        # t_evals holds (t_start, t_end). We just need t_end for our analytical formula
        t_end = t_evals[..., 1]
        
        # Hardcode the known physics parameters for ideal Ramsey 
        A = 0.5
        B = 0.5
        # The detuning frequency defined in your piecewise Hamiltonian was ~ 0.005 * freq
        omega_m = jnp.pi * qubit_info.frequency * 0.005
        phi = 0.0
        
        # Calculate closed-form expectation value directly
        prob_minus = A * jnp.exp(-t_end / T2) * jnp.cos(omega_m * t_end + phi) + B
        
        # Clip to ensure numerical stability during optimization
        prob_minus = jnp.clip(prob_minus, 1e-5, 1.0 - 1e-5)

        numpyro.sample("shots", dist.BinomialProbs(total_count=1000, probs=prob_minus))
```

```python
import jax
import jax.numpy as jnp
from jax import jacfwd

# 1. Define your physical model (e.g., a decaying signal)
def model(theta, x):
    # theta: parameters [amplitude, decay_rate]
    # x: design point (time of measurement)
    return theta[0] * jnp.exp(-theta[1] * x)

# 2. Define the Laplace EIG function
def estimate_laplace_eig(design_x, prior_cov, noise_var, current_theta):
    """
    Estimates EIG at a specific design point using the Fisher Information.
    """
    # Calculate the Jacobian of the model w.r.t parameters theta
    # J is the sensitivity of the measurement to parameter changes
    jacobian_func = jacfwd(lambda t: model(t, design_x))
    J = jacobian_func(current_theta) # Shape: (1, n_params)
    
    # Reshape J if it's a single measurement
    J = J.reshape(1, -1)
    
    # 3. Calculate the Fisher Information Matrix (FIM)
    # FIM = (1/noise_sigma^2) * J^T * J
    precision_matrix = (1.0 / noise_var) * (J.T @ J)
    
    # 4. Calculate EIG = 0.5 * log|I + PriorCov * FIM|
    # This represents the expected reduction in entropy
    dim = prior_cov.shape[0]
    term_to_det = jnp.eye(dim) + prior_cov @ precision_matrix
    eig = 0.5 * jnp.log(jnp.linalg.det(term_to_det))
    
    return eig

# --- Execution ---
theta_guess = jnp.array([1.0, 0.1])    # Current best estimate of parameters
prior_covariance = jnp.diag(jnp.array([0.1, 0.01])) # Uncertainty in params
measurement_noise = 0.05               # Known sensor noise variance

# Test design points (e.g., measuring at t=1s vs t=50s)
for t in [1.0, 10.0, 50.0]:
    val = estimate_laplace_eig(t, prior_covariance, measurement_noise, theta_guess)
    print(f"EIG at time {t}s: {val:.4f}")
```

```python
import jax
import jax.numpy as jnp

# 1. Define your physical model (e.g., a decaying signal)
def model(theta, x):
    # theta: parameters [amplitude, decay_rate]
    # x: design point (time of measurement)
    return theta[0] * jnp.exp(-theta[1] * x)

# 2. Define the Laplace EIG function
def estimate_laplace_eig(design_x, prior_cov, noise_var, current_theta):
    """
    Estimates EIG at a specific design point using the Fisher Information.
    """
    # Calculate the Jacobian of the model w.r.t parameters theta
    # J is the sensitivity of the measurement to parameter changes
    jacobian_func = jax.jacfwd(lambda t: model(t, design_x))

    J = jacobian_func(current_theta) # Shape: (1, n_params)
    
    # Reshape J if it's a single measurement
    J = J.reshape(1, -1)
    
    # 3. Calculate the Fisher Information Matrix (FIM)
    # FIM = (1/noise_sigma^2) * J^T * J
    precision_matrix = (1.0 / noise_var) * (J.T @ J)
    
    # 4. Calculate EIG = 0.5 * log|I + PriorCov * FIM|
    # This represents the expected reduction in entropy
    dim = prior_cov.shape[0]
    term_to_det = jnp.eye(dim) + prior_cov @ precision_matrix
    eig = 0.5 * jnp.log(jnp.linalg.det(term_to_det))
    
    return eig

# --- Execution ---
theta_guess = jnp.array([1.0, 0.1])    # Current best estimate of parameters
prior_covariance = jnp.diag(jnp.array([0.1, 0.01])) # Uncertainty in params
measurement_noise = 0.05               # Known sensor noise variance

# Test design points (e.g., measuring at t=1s vs t=50s)
for t in [1.0, 10.0, 50.0]:
    val = estimate_laplace_eig(t, prior_covariance, measurement_noise, theta_guess)
    print(f"EIG at time {t}s: {val:.4f}")
```

```python
import jax
import jax.numpy as jnp

def estimate_T2_eig(current_T2_guess, design_param, prior_var_T2):
    """
    Estimates EIG for a specific T2 value and a specific timing design.
    
    Args:
        current_T2_guess: float, our current best estimate of T2.
        design_param: dict, contains 'start' and 'end' times.
        prior_var_T2: float, the variance of our current belief about T2.
        noise_var: float, the readout noise variance of the qubit.
    """
    
    # 1. Calculate the gradient (Jacobian) of the model w.r.t T2
    # We use jax.grad because device_with_T2 returns a scalar (float)
    grad_func = jax.value_and_grad(device_with_T2, argnums=0)
    value, sensitivity = grad_func(current_T2_guess, ravel_fn(design_param))
    
    n = 1000  # Number of shots, can be adjusted based on your experiment
    noise_var = (1/n) * (1 - value**2)  # Variance of a Bernoulli with p = (1 + value) / 2
    # 2. Calculate Fisher Information (scalar version)
    # F = (df/dT2)^2 / sigma^2
    fisher_info = (sensitivity**2) / noise_var
    
    # 3. Calculate EIG using the Laplace-Bayes formula
    # EIG = 0.5 * log(1 + PriorVariance * FisherInformation)
    # This represents the log-ratio of the prior uncertainty vs posterior uncertainty
    eig = 0.5 * jnp.log(1 + prior_var_T2 * fisher_info)
    
    return eig

# Setup your priors and noise
current_T2 = 475.  # Assume T2 is 50 microseconds
prior_variance = 100.0  # High uncertainty in T2 (std dev of 10 microseconds)
readout_noise = 0.01  # Noise in the <Z> measurement

# Define a range of times to test (e.g., 1us to 200us)
test_times = jnp.linspace(300, 1000, 201)  # 10 candidate measurement times between 300us and 1000us
eigs = []

key = jax.random.key(0)

for t_end in test_times:
    # Update the design dictionary
    test_param = {
        "start": {"t": 0.0},
        "end": {"t": t_end},
        # ... include any other required params for your hamiltonian/unravel_fn
    }

    key, subkey = jax.random.split(key)
    T2_samples = jax.random.normal(subkey, shape=(100, )) * jnp.sqrt(prior_variance) + current_T2
    
    vals = jax.vmap(estimate_T2_eig, in_axes=(0, None, None))(T2_samples, test_param, prior_variance)
    val = jnp.mean(vals)
    eigs.append(val)

# Find the time that maximizes EIG
best_idx = jnp.argmax(jnp.array(eigs))
print(f"Optimal measurement time: {test_times[best_idx] * 1e6:.2f} microseconds")
```

```python
import jax
import jax.numpy as jnp
from jax import jacfwd


# 1. Model with Multiple Observables and Multiple Parameters
def model_multi_obs(theta, x):
    T2, Omega = theta[0], theta[1]

    # Let's say we measure both <X> and <Y> expectations
    # <X> oscillates with cos, <Y> oscillates with sin
    decay = jnp.exp(-x / T2)
    obs_X = decay * jnp.cos(Omega * x)
    obs_Y = decay * jnp.sin(Omega * x)

    # Return a vector of observables
    return jnp.array([obs_X, obs_Y])


# 2. The Updated Laplace EIG Function
def estimate_multi_obs_eig(theta_vector, design_x, prior_cov_matrix, noise_cov_matrix):
    # 1. Use jacfwd because the output is a vector
    # J shape will be (num_observables, num_parameters)
    jacobian_func = jacfwd(model_multi_obs, argnums=0)
    J = jacobian_func(theta_vector, design_x)

    # 2. Calculate the FIM using the noise covariance matrix
    # F = J^T * NoiseInv * J
    noise_inv = jnp.linalg.inv(noise_cov_matrix)
    FIM = J.T @ noise_inv @ J

    # 3. Calculate EIG exactly as before
    dim = prior_cov_matrix.shape[0]
    term_to_det = jnp.eye(dim) + prior_cov_matrix @ FIM
    eig = 0.5 * jnp.log(jnp.linalg.det(term_to_det))

    return eig


# --- Example Setup ---
current_theta = jnp.array([50.0e-6, 1e6])  # [T2, Omega]
prior_cov = jnp.diag(jnp.array([(10e-6) ** 2, (0.1e6) ** 2]))

# Assume 5% noise on both X and Y, and the noise is uncorrelated
noise_variance = 0.05**2
noise_cov = jnp.diag(jnp.array([noise_variance, noise_variance]))

design_time = 10e-6
eig_val = estimate_multi_obs_eig(current_theta, design_time, prior_cov, noise_cov)
```

## Parameter Characterization

As the $T_1$ experiment, we simply transform the data by ourself. Now we can fit the function using the following exponential decay with cosine oscilation model,

$$
    \rm{model}(t) = A e^{-t / T_{2}} \cos(\omega_{m} t + \phi) + B
$$

Reference: https://qiskit-community.github.io/qiskit-experiments/manuals/characterization/t2ramsey.html

```python
import optax

expval_z = exp_data.get_observed().reshape(-1)
t_range = exp_data.get_parameter()[:, -1]


def exponential_decay_cosine_model(t, A, B, omega_m, T2, phi):
    return A * jnp.exp(-t / T2) * jnp.cos(omega_m * t + phi) + B


def loss_fn(param):
    # The data is from each discrete t.
    excited_state_prob = sq.utils.expectation_value_to_prob_minus(expval_z)

    predicted = exponential_decay_cosine_model(
        t_range, param["A"], param["B"], param["omega_m"], param["T2"], param["phi"]
    )
    cost = sq.models.shared.mse(excited_state_prob, predicted)
    return cost, {"mse": cost}


init_param = {
    "T2": jnp.array(450.0),
    "omega_m": jnp.array(jnp.pi) * qubit_info.frequency * 0.005,
    "A": jnp.array(0.8),
    "B": jnp.array(0.5),
    "phi": jnp.array(0.0),
}

iterations = 100
optimized_param, hist = sq.optimize.minimize(
    init_param,
    loss_fn,
    optax.adam(1e-3),
    maxiter=iterations,
)

fited = exponential_decay_cosine_model(t_range, **optimized_param)  # type: ignore
hist[-1]
```

Alternatively, we can optimize using other optimization algorithm from other library. In the below example, we use `jax.scipy.optimize` with a thin wrapper of the original functions above.

```python
import jax.scipy.optimize


def unravel_fn2(param):
    return {
        "T2": param[0],
        "omega_m": param[1],
        "A": param[2],
        "B": param[3],
        "phi": param[4],
    }


def wrapper(fn):
    def _fn(param):
        unflatted_param = unravel_fn2(param)

        loss, _ = fn(unflatted_param)

        return loss

    return _fn


res = jax.scipy.optimize.minimize(
    wrapper(loss_fn), x0=jnp.array(list(init_param.values())), method="BFGS"
)

fited = exponential_decay_cosine_model(t_range, **unravel_fn2(res.x))  # type: ignore
unravel_fn2(res.x)
```

Even fancier, we can do bayesian inference for the model parameters.

```python
import numpyro
from numpyro.infer import MCMC, NUTS, Predictive, init_to_sample
```

```python
def probabilistic_model(t_eval, observed: None = None):
    A = numpyro.sample("A", numpyro.distributions.Normal(0.8, 0.05))
    B = numpyro.sample("B", numpyro.distributions.Normal(0.5, 0.05))
    omega_m = numpyro.sample(
        "omega_m",
        numpyro.distributions.Normal(jnp.pi * qubit_info.frequency * 0.005, 0.001),
    )
    T2 = numpyro.sample("T2", numpyro.distributions.Normal(500.0, 20.0))
    phi = numpyro.sample("phi", numpyro.distributions.Normal(0.0, 0.01))

    predicted = exponential_decay_cosine_model(t_eval, A, B, omega_m, T2, phi)

    numpyro.sample(
        "population",
        numpyro.distributions.TruncatedNormal(predicted, 0.05, low=0.0, high=1.0),  # type: ignore
        obs=observed,
    )


mcmc = MCMC(
    NUTS(probabilistic_model, init_strategy=init_to_sample),
    num_warmup=1000,
    num_samples=1000,
)
mcmc.run(jax.random.key(0), t_range, sq.utils.expectation_value_to_prob_minus(expval_z))

samples = mcmc.get_samples()
population_samples = Predictive(
    probabilistic_model, posterior_samples=samples, num_samples=1000
)(jax.random.key(0), t_range)["population"]

mcmc.print_summary()
```

```python
mean_pred = np.mean(population_samples, axis=0)
std_pred = np.std(population_samples, axis=0)

plt.plot(t_range, mean_pred, "-", label="Mean", color=sq.utils.colors.blue)
plt.fill_between(
    t_range,
    mean_pred - std_pred,
    mean_pred + std_pred,
    color=sq.utils.colors.blue,
    alpha=0.3,
    label="±1 std. dev.",
)

plt.plot(
    t_range,
    sq.utils.expectation_value_to_prob_minus(expval_z),
    ".",
    label="Data",
    color=sq.utils.colors.red,
)

plt.legend()
```
